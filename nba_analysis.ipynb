# NBA analysis

Loads and cleans the data with `nba.data_wrangling`, then draws the charts with `nba.visualisation`.
Make sure the NBA dataset has been downloaded as described in the README.

```python
import sqlite3
from pathlib import Path

from nba.data_wrangling import load_draft, load_games
from nba.visualisation import (
    plot_draftee_origin,
    plot_home_win_percentage,
    plot_top5_picks_by_team,
)

DATA_PATH = Path.cwd() / "data"
con = sqlite3.connect(DATA_PATH / "nba.sqlite")
```

## Load and clean the data

```python
games = load_games(con)
draft = load_draft(con)
games.head()
```

## Home team win percentage by year

```python
plot_home_win_percentage(games)
```

## Where draftees came from, by decade

```python
plot_draftee_origin(draft)
```

## Teams with the most top-5 draft picks

```python
plot_top5_picks_by_team(draft)
```
